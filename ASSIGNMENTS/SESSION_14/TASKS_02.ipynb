#### Task 2: Expected counts under independence - music streaming preference by gender.

```python
import numpy as np
import pandas as pd
 
observed = pd.DataFrame({"Males": [12, 8, 10], "Females": [18, 7, 15]},
                        index=["Spotify", "Gaana", "YouTube Music"])
 
row_totals = observed.sum(axis=1)          # per music app
col_totals = observed.sum(axis=0)          # per gender
grand_total = observed.values.sum()
 
print("Observed counts with totals:")
with_totals = observed.copy()
with_totals["Total"] = row_totals
with_totals.loc["Total"] = list(col_totals) + [grand_total]
print(with_totals.to_string(), "\n")
 
# Expected count = (row total x column total) / grand total
expected = pd.DataFrame(np.outer(row_totals, col_totals) / grand_total,
                        index=observed.index, columns=observed.columns)
 
print("Step-by-step:")
for app in observed.index:
    for g in observed.columns:
        print(f"  E({app}, {g}) = ({row_totals[app]} x {col_totals[g]}) / {grand_total} = {expected.loc[app, g]:.3f}")
 
print("\nExpected counts assuming independence:")
print(expected.round(3).to_string())
print(f"\nCheck: expected totals = {expected.values.sum():.1f} (equals grand total {grand_total})")
```
