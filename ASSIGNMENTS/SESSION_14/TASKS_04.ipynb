#### Task 4: Chi-square goodness of fit - does 70% of users preferring the new Flipkart product hold?

```python
import numpy as np
import pandas as pd
from scipy import stats
 
alpha = 0.05
n = 30
observed = np.array([18, 12])                 # prefer new product, do not prefer
claimed = np.array([0.70, 0.30])
labels = ["Prefer new product", "Do not prefer"]
 
print("H0: the proportions are 70% prefer / 30% do not prefer (the company's claim is true)")
print("H1: the proportions differ from 70% / 30% (the claim is false)\n")
 
expected = n * claimed                         # expected counts from the claimed proportions
contrib = (observed - expected) ** 2 / expected
table = pd.DataFrame({"Observed": observed, "Claimed %": claimed * 100,
                      "Expected": expected, "(O-E)^2/E": contrib}, index=labels)
print(table.round(3).to_string(), "\n")
 
chi2 = contrib.sum()
df = len(observed) - 1
critical = stats.chi2.ppf(1 - alpha, df)
p_value = stats.chi2.sf(chi2, df)
print(f"chi-square = {contrib[0]:.4f} + {contrib[1]:.4f} = {chi2:.4f}")
print(f"degrees of freedom = categories - 1 = {df}")
print(f"critical value (alpha = {alpha}) = {critical:.3f}")
print(f"p-value = {p_value:.4f}")
print(f"All expected counts >= 5: {bool((expected >= 5).all())}\n")
 
chi2_s, p_s = stats.chisquare(observed, f_exp=expected)
print(f"scipy.stats.chisquare check: chi2 = {chi2_s:.4f}, p = {p_s:.4f}\n")
 
if chi2 > critical:
    print("Decision: reject H0 - the data contradict the 70% claim.")
else:
    print("Decision: fail to reject H0 - the data do not contradict the claim that 70% prefer the new product.")
print(f"Observed preference = {observed[0] / n:.0%} (claimed 70%)")
```
