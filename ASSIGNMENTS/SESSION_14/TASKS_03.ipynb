#### Task 3: Chi-square test of independence on the Task 1 data (favourite app vs payment method).
Run task1_contingency_table.py first so survey_responses.csv exists.

```python
import os
import numpy as np
import pandas as pd
from scipy import stats
 
HERE = (os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd())
survey = pd.read_csv(os.path.join(HERE, "survey_responses.csv"))
apps = ["Zomato", "Swiggy", "Uber Eats"]
payments = ["Paytm", "PhonePe", "Credit Card"]
alpha = 0.05
 
observed = pd.crosstab(survey["favorite_app"], survey["payment_method"]).loc[apps, payments]
print("H0: favourite app and payment method are independent (no association)")
print("H1: favourite app and payment method are associated\n")
print("Step 1 - Observed counts:")
print(observed.to_string(), "\n")
 
# Step 2 - expected counts
row_t, col_t, n = observed.sum(axis=1), observed.sum(axis=0), observed.values.sum()
expected = pd.DataFrame(np.outer(row_t, col_t) / n, index=apps, columns=payments)
print("Step 2 - Expected counts = row total x column total / grand total:")
print(expected.round(2).to_string())
print(f"Smallest expected count = {expected.values.min():.2f} (rule of thumb: should be >= 5)\n")
 
# Step 3 - contribution of each cell (O - E)^2 / E
contrib = (observed - expected) ** 2 / expected
print("Step 3 - Cell contributions (O - E)^2 / E:")
print(contrib.round(3).to_string(), "\n")
 
# Step 4 - test statistic, degrees of freedom, critical value, p-value
chi2 = contrib.values.sum()
df = (observed.shape[0] - 1) * (observed.shape[1] - 1)
critical = stats.chi2.ppf(1 - alpha, df)
p_value = stats.chi2.sf(chi2, df)
print(f"Step 4 - chi-square = {chi2:.3f}, degrees of freedom = (3-1)(3-1) = {df}")
print(f"         critical value at alpha = {alpha}: {critical:.3f}")
print(f"         p-value = {p_value:.4f}\n")
```
