#### Task 1: Contingency table - favourite food delivery app vs preferred payment method (60 survey responses).

```python
import os
import numpy as np
import pandas as pd

HERE = (os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd())  # directory of this script
rng = np.random.default_rng(seed=1)

apps = ["Zomato", "Swiggy", "Uber Eats"]
payments = ["Paytm", "PhonePe", "Credit Card"]

# Survey sample (simulated, 60 respondents): counts of each (app, payment) combination
counts = {
    ("Zomato", "Paytm"): 9,     ("Zomato", "PhonePe"): 7,     ("Zomato", "Credit Card"): 6,
    ("Swiggy", "Paytm"): 5,     ("Swiggy", "PhonePe"): 10,    ("Swiggy", "Credit Card"): 5,
    ("Uber Eats", "Paytm"): 5,  ("Uber Eats", "PhonePe"): 6,  ("Uber Eats", "Credit Card"): 7,
}
rows = [(a, p) for (a, p), c in counts.items() for _ in range(c)]
survey = pd.DataFrame(rows, columns=["favorite_app", "payment_method"])
survey = survey.sample(frac=1, random_state=1).reset_index(drop=True)      # shuffle respondents
survey.insert(0, "respondent_id", [f"R{i:02d}" for i in range(1, len(survey) + 1)])
survey.to_csv(os.path.join(HERE, "survey_responses.csv"), index=False)

print(f"Number of responses: {len(survey)}\n")
print("First 5 responses:")
print(survey.head().to_string(index=False), "\n")

# Contingency table (cross-tabulation) with row and column totals
table = pd.crosstab(survey["favorite_app"], survey["payment_method"],
                    margins=True, margins_name="Total")
table = table.loc[apps + ["Total"], payments + ["Total"]]
print("Contingency table (rows = favourite app, columns = payment method):")
print(table.to_string())

print("\nRow percentages (payment mix within each app):")
print((pd.crosstab(survey["favorite_app"], survey["payment_method"], normalize="index")
       .loc[apps, payments] * 100).round(1).to_string())
```
