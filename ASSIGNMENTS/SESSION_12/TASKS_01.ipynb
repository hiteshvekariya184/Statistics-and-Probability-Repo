##### Task 1: Null and alternative hypotheses - did the cashback offer increase average daily orders?

```python
import numpy as np
import pandas as pd
from scipy import stats

rng = np.random.default_rng(seed=10)

print("Hypotheses (mu = average number of daily orders)")
print("  H0: mu_after <= mu_before   (the cashback offer did not increase daily orders)")
print("  H1: mu_after >  mu_before   (the cashback offer increased daily orders)  -> one-tailed test\n")

# 30 days of orders before and 30 days after the offer (simulated)
before = rng.normal(loc=1200, scale=90, size=30).round()
after = rng.normal(loc=1260, scale=90, size=30).round()
daily = pd.DataFrame({"before": before, "after": after})
print(daily.describe().loc[["count", "mean", "std"]].round(1).to_string(), "\n")

alpha = 0.05
t_stat, p_value = stats.ttest_ind(after, before, equal_var=False, alternative="greater")
print(f"Welch t-test (one-tailed): t = {t_stat:.3f}, p-value = {p_value:.4f}, alpha = {alpha}")
if p_value < alpha:
    print("Decision: reject H0 -> evidence that the cashback offer increased average daily orders.")
else:
    print("Decision: fail to reject H0 -> not enough evidence that the offer increased daily orders.")
```
