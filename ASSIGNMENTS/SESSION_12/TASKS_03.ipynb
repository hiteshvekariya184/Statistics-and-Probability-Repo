#### Task 3: p-value = 0.03 at 95% confidence -> decision, plus a worked A/B test on Flipkart-style data.

```python
import numpy as np
import pandas as pd
from scipy import stats

def decide(p_value, confidence=0.95):
    alpha = 1 - confidence
    verdict = "REJECT H0" if p_value < alpha else "FAIL TO REJECT H0"
    return alpha, verdict

# ---- Part A: the given numbers ----
p_given = 0.03
alpha, verdict = decide(p_given, 0.95)
print(f"Given p-value = {p_given}, confidence = 95% -> alpha = {alpha:.2f}")
print(f"Compare p ({p_given}) with alpha ({alpha:.2f}): {p_given} < {alpha:.2f} -> {verdict}\n")

# ---- Part B: an A/B test that produces a p-value close to 0.03 ----
visitors = 2000
control_buy, banner_buy = 200, 237              # purchases without / with the discount banner
p_c, p_b = control_buy / visitors, banner_buy / visitors
pooled = (control_buy + banner_buy) / (2 * visitors)
se = np.sqrt(pooled * (1 - pooled) * (2 / visitors))
z = (p_b - p_c) / se
p_value = 1 - stats.norm.cdf(z)                 # one-tailed: banner INCREASES purchases

print("A/B test (H0: banner conversion <= control, H1: banner conversion > control)")
print(pd.DataFrame({"Visitors": [visitors, visitors], "Purchases": [control_buy, banner_buy],
                    "Conversion %": [p_c * 100, p_b * 100]}, index=["Control", "Discount banner"]).round(2).to_string())
print(f"\nz = {z:.3f}, one-tailed p-value = {p_value:.4f}")
alpha, verdict = decide(p_value, 0.95)
print(f"Decision at 95% confidence (alpha = {alpha:.2f}): {verdict}")
print(f"Lift: {p_b - p_c:.2%} points ({(p_b / p_c - 1):.1%} relative)")
```
