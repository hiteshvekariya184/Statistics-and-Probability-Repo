#### Task 2: Type I vs Type II error - music app claims listening time rises from 40 to 45 minutes.

```python
import numpy as np
import pandas as pd
from scipy import stats

rng = np.random.default_rng(seed=20)

print("H0: mean listening time = 40 min (algorithm does not work)")
print("H1: mean listening time > 40 min (algorithm works)\n")

alpha, n, sd, trials = 0.05, 50, 12, 10_000

def rejection_rate(true_mean):
    """Share of repeated experiments in which H0 gets rejected."""
    rejected = 0
    for _ in range(trials):
        sample = rng.normal(true_mean, sd, n)
        p = stats.ttest_1samp(sample, popmean=40, alternative="greater").pvalue
        rejected += p < alpha
    return rejected / trials

type1 = rejection_rate(40)      # H0 is TRUE (algorithm useless) -> any rejection is a false alarm
power = rejection_rate(45)      # H0 is FALSE (algorithm really adds 5 min) -> rejection is correct
type2 = 1 - power

table = pd.DataFrame({
    "Reality": ["Algorithm does NOT work (mean = 40)", "Algorithm works (mean = 45)"],
    "Reject H0 (say 'it works')": [f"Type I error  ({type1:.3f})", f"Correct, power ({power:.3f})"],
    "Fail to reject H0": [f"Correct ({1 - type1:.3f})", f"Type II error ({type2:.3f})"],
})
print(table.to_string(index=False), "\n")
print(f"Simulated Type I error rate  = {type1:.3f}  (close to alpha = {alpha})")
print(f"Simulated Type II error rate = {type2:.3f}")
print("\nConcluding that the algorithm works when it actually does not is a TYPE I ERROR (false positive).")
```
