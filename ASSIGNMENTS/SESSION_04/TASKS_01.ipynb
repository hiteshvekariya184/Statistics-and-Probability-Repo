## Task 1: Range of Daily Step Counts

```python
import numpy as np
import pandas as pd
 
steps = pd.Series([6500, 7200, 8000, 6000, 9000, 7500, 8200], name="steps")
 
# Range = Maximum value - Minimum value
step_range = steps.max() - steps.min()
 
print("Daily steps :", steps.tolist())
print(f"Maximum     : {steps.max()}")
print(f"Minimum     : {steps.min()}")
print(f"Range       : {step_range}")
print(f"Range (numpy ptp): {np.ptp(steps)}")
 
print()
print("Interpretation:")
print(f"The range is {step_range} steps, so the gap between the most active day")
print(f"({steps.max()}) and the least active day ({steps.min()}) is fairly wide.")
print("The user stays active every day (never below 6000 steps), but activity")
print("level changes noticeably from day to day rather than being steady.")
```
