## Task 5: Variance and Standard Deviation of AI-Generated UPI Data

```python
import numpy as np
import pandas as pd
 
upi = pd.Series([150, 220, 95, 480, 130, 310, 175, 260, 205], name="upi_amount")
 
mean = upi.mean()
variance = np.var(upi, ddof=0)
sd = np.std(upi, ddof=0)
cv = (sd / mean) * 100
 
print("UPI amounts (Rs.)  :", upi.tolist())
print(f"Mean               : Rs.{mean:.2f}")
print(f"Variance           : {variance:.2f}")
print(f"Standard deviation : Rs.{sd:.2f}")
print(f"CV                 : {cv:.2f}%")
 
print()
print("Interpretation:")
print(f"The standard deviation (Rs.{sd:.2f}) is about {cv:.0f}% of the mean, which is high.")
print("Spending is not very stable: most days are small payments, but a few")
print("larger transactions (for example Rs.480) push the spread up.")
```
