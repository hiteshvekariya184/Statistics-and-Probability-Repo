## Task 4: Standard Deviation vs Mean of Instagram Likes

```python
import numpy as np
import pandas as pd
 
likes = pd.Series([120, 150, 130, 110, 170, 140, 135, 160, 125, 155], name="likes")
 
mean = likes.mean()
variance = np.var(likes, ddof=0)
sd = np.std(likes, ddof=0)
ratio = (sd / mean) * 100
 
print("Likes              :", likes.tolist())
print(f"Mean               : {mean:.2f}")
print(f"Variance           : {variance:.2f}")
print(f"Standard deviation : {sd:.2f}")
print(f"SD as % of mean    : {ratio:.2f}%")
 
print()
print("Comment:")
print(f"The standard deviation ({sd:.2f}) is only about {ratio:.1f}% of the mean ({mean:.1f}).")
print("Likes stay in a fairly narrow band around the average, so the engagement")
print("is reasonably consistent across posts, with no extreme spikes or drops.")
```
