#### Task 3: Coefficient of Variation (CV) of Monthly Mobile Data Usage
CV = (Standard Deviation / Mean) x 100%

```python
import numpy as np
import pandas as pd
from scipy import stats
 
usage = pd.Series([3.2, 2.8, 3.5, 3.0, 2.9, 3.1], name="data_gb")
 
mean = usage.mean()
sd = np.std(usage, ddof=0)            # population standard deviation
cv = (sd / mean) * 100
 
print("Monthly usage (GB):", usage.tolist())
print(f"Mean               : {mean:.3f} GB")
print(f"Standard deviation : {sd:.3f} GB")
print(f"Coefficient of Var : {cv:.2f}%")
print(f"CV (scipy.stats.variation): {stats.variation(usage) * 100:.2f}%")
 
print()
print("Interpretation:")
if cv < 15:
    print(f"CV is {cv:.2f}% (below 15%), so the user's data usage is consistent.")
else:
    print(f"CV is {cv:.2f}% (15% or more), so the user's data usage is highly variable.")
```
