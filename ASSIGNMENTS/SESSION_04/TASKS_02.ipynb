## Task 2: Variance and Standard Deviation of Zomato Order Amounts

```python
import numpy as np
import pandas as pd
 
orders = pd.Series([250, 400, 350, 300, 500], name="order_amount")
 
print("Order amounts (Rs.):", orders.tolist())
print(f"Mean                     : Rs.{orders.mean():.2f}")
 
# Population (divide by n)
print(f"\nPopulation variance      : {np.var(orders, ddof=0):.2f}")
print(f"Population std deviation : Rs.{np.std(orders, ddof=0):.2f}")
 
# Sample (divide by n - 1)
print(f"\nSample variance          : {orders.var(ddof=1):.2f}")
print(f"Sample std deviation     : Rs.{orders.std(ddof=1):.2f}")
```
