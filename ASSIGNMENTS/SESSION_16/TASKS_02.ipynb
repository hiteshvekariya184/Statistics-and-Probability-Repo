#### Task 2: Spearman rank correlation between IPL players' total runs and Instagram followers.

```python
import os
import pandas as pd
from scipy import stats
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
 
# Illustrative data for 12 anonymised players (not real statistics)
players = pd.DataFrame({
    "player": [f"Player {i:02d}" for i in range(1, 13)],
    "total_runs": [7000, 6500, 5800, 5200, 4800, 4300, 3900, 3500, 3000, 2600, 2200, 1800],
    "followers_millions": [40, 55, 12, 30, 8, 22, 15, 9, 18, 5, 7, 3],
})
players.to_csv(os.path.join(HERE, "ipl_players.csv"), index=False)
print(players.to_string(index=False), "\n")
 
rho, p_value = stats.spearmanr(players["total_runs"], players["followers_millions"])
print(f"Spearman correlation (rho) = {rho:.4f}")
print(f"p-value                    = {p_value:.4f}")
 
# For comparison: Pearson on the raw values
r = players["total_runs"].corr(players["followers_millions"])
print(f"Pearson r on raw values    = {r:.4f}\n")
 
print("Interpretation:")
print(f"  rho = {rho:.2f} is a strong positive rank correlation: players with more runs tend to rank higher in followers.")
print(f"  p = {p_value:.4f} < 0.05, so the relationship is statistically significant in this sample.")
print("  It is not perfect (rho < 1): e.g. Player 03 has 5800 runs but only 12M followers, Player 09 has 3000 runs but 18M.")
print("  So higher runs GENERALLY go with more followers, but runs alone do not determine followers.")
```
