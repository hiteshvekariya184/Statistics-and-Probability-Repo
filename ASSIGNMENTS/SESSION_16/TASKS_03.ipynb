#### Task 3: Correlation matrix of Flipkart product features: price, user rating, number of reviews.

```python
import os
import numpy as np
import pandas as pd
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
rng = np.random.default_rng(seed=3)
 
# Simulated data for 30 products
n = 30
price = rng.integers(300, 5000, n)
rating = np.clip(3.4 + 0.00005 * price + rng.normal(0, 0.35, n), 1, 5).round(1)
reviews = np.clip(2500 - 0.35 * price + 450 * (rating - 3.5) + rng.normal(0, 300, n), 20, None).round()
products = pd.DataFrame({"product_id": [f"P{i:02d}" for i in range(1, n + 1)],
                         "price": price, "user_rating": rating, "num_reviews": reviews.astype(int)})
products.to_csv(os.path.join(HERE, "flipkart_products.csv"), index=False)
 
print("First 5 products:")
print(products.head().to_string(index=False), "\n")
 
corr_matrix = products[["price", "user_rating", "num_reviews"]].corr()      # Pearson by default
print("Correlation matrix:")
print(corr_matrix.round(3).to_string())
 
print("\nReading the matrix:")
for a, b in [("price", "user_rating"), ("price", "num_reviews"), ("user_rating", "num_reviews")]:
    print(f"  {a} vs {b}: r = {corr_matrix.loc[a, b]:.3f}")
```
