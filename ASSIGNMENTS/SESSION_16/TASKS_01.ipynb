#### Task 1: Create songs.csv (12 songs) and compute Pearson correlation between danceability and popularity.

```python
import os
import pandas as pd
from scipy import stats
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
 
# Sample dataset created manually (illustrative values, fictional tracks)
songs = pd.DataFrame({
    "song": [f"Track {i:02d}" for i in range(1, 13)],
    "danceability": [0.82, 0.65, 0.71, 0.54, 0.90, 0.48, 0.77, 0.60, 0.86, 0.42, 0.69, 0.58],
    "energy":       [0.75, 0.60, 0.68, 0.80, 0.85, 0.45, 0.70, 0.55, 0.88, 0.50, 0.62, 0.72],
    "popularity":   [74, 55, 70, 62, 85, 50, 66, 48, 78, 52, 68, 45],
})
path = os.path.join(HERE, "songs.csv")
songs.to_csv(path, index=False)
songs = pd.read_csv(path)          # read it back, as you would with a downloaded dataset
 
print(songs.to_string(index=False), "\n")
 
# Pearson correlation using pandas
r = songs["danceability"].corr(songs["popularity"])          # method="pearson" is the default
print(f"Pearson correlation (danceability, popularity) = {r:.4f}")
 
# Optional check: significance with scipy
r_s, p = stats.pearsonr(songs["danceability"], songs["popularity"])
print(f"scipy check: r = {r_s:.4f}, p-value = {p:.5f}")
 
strength = "very strong" if abs(r) >= 0.8 else "strong" if abs(r) >= 0.6 else "moderate" if abs(r) >= 0.4 else "weak"
print(f"\nInterpretation: a {strength} {'positive' if r > 0 else 'negative'} linear relationship "
      f"(r^2 = {r ** 2:.2f}, \nso danceability explains about {r ** 2:.0%} of the variation in popularity in this sample).")
```
