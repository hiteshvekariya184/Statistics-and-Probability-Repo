#### Task 4: Seaborn heatmap of the Flipkart correlation matrix (annotations + colour bar).
Run task3_flipkart_correlation_matrix.py first so flipkart_products.csv exists.

```python
import os
import pandas as pd
import matplotlib
if "__file__" in globals():            # plain script: save to file; notebook: show inline
    matplotlib.use("Agg")
import matplotlib.pyplot as plt
import seaborn as sns
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
products = pd.read_csv(os.path.join(HERE, "flipkart_products.csv"))
corr_matrix = products[["price", "user_rating", "num_reviews"]].corr()
 
plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
            linewidths=0.5, square=True, cbar=True, cbar_kws={"label": "Pearson correlation (r)"})
plt.title("Flipkart products: correlation heatmap")
plt.tight_layout()
plt.savefig(os.path.join(HERE, "heatmap.png"), dpi=150)
if "__file__" not in globals():
    plt.show()
print("Heatmap saved to heatmap.png")
print(corr_matrix.round(2).to_string())
```
