#### Task 1: One-sample t-test - is the average Zomato delivery time different from 30 minutes?

```python
import numpy as np
from scipy import stats
 
times = np.array([28, 32, 27, 35, 29, 31, 30, 33, 26, 34])
mu0, alpha = 30, 0.05
 
print("H0: mean delivery time = 30 minutes")
print("H1: mean delivery time != 30 minutes  (two-tailed test)\n")
 
n = len(times)
mean, sd = times.mean(), times.std(ddof=1)
se = sd / np.sqrt(n)
print(f"n = {n}, sample mean = {mean:.2f} min, sample SD = {sd:.3f}, standard error = {se:.3f}")
 
t_stat, p_value = stats.ttest_1samp(times, popmean=mu0)       # two-sided by default
df = n - 1
ci = stats.t.interval(0.95, df, loc=mean, scale=se)
t_crit = stats.t.ppf(1 - alpha / 2, df)
 
print(f"t-statistic = {t_stat:.3f}  (degrees of freedom = {df})")
print(f"p-value     = {p_value:.4f}")
print(f"Critical t (alpha = 0.05, two-tailed) = +/- {t_crit:.3f}")
print(f"95% confidence interval for the mean = ({ci[0]:.2f}, {ci[1]:.2f}) minutes\n")
 
if p_value < alpha:
    print("Decision: reject H0 - average delivery time is significantly different from 30 minutes.")
else:
    print("Decision: fail to reject H0 - no significant evidence that the average delivery time differs from 30 minutes.")
```
