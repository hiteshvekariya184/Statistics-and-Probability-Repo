#### Task 3: Paired t-test - screen time of 10 users before and after enabling Focus Mode.

```python
import numpy as np
import pandas as pd
from scipy import stats
 
rng = np.random.default_rng(seed=3)
alpha = 0.05
 
# SAME 10 users measured twice (simulated), hours per day
before = rng.normal(loc=5.5, scale=1.0, size=10).round(1)
after = (before - rng.normal(loc=0.7, scale=0.5, size=10)).round(1)
 
data = pd.DataFrame({"user": [f"U{i}" for i in range(1, 11)], "before_hrs": before,
                     "after_hrs": after})
data["difference"] = (data["before_hrs"] - data["after_hrs"]).round(1)
print(data.to_string(index=False), "\n")
 
print("H0: mean(before - after) <= 0   (Focus Mode does not reduce screen time)")
print("H1: mean(before - after) >  0   (Focus Mode reduces screen time)   (one-tailed)\n")
 
t_stat, p_value = stats.ttest_rel(before, after, alternative="greater")
d = data["difference"]
cohens_d = d.mean() / d.std(ddof=1)
 
print(f"Mean before = {before.mean():.2f} h, mean after = {after.mean():.2f} h, "
      f"average reduction = {d.mean():.2f} h")
print(f"ttest_rel(before, after, alternative='greater'): t = {t_stat:.3f}, "
      f"df = {len(d) - 1}, p-value = {p_value:.5f}")
print(f"Effect size (Cohen's d for paired data) = {cohens_d:.2f}")
 
if p_value < alpha:
    print("\nDecision: reject H0 - Focus Mode significantly reduced screen time.")
else:
    print("\nDecision: fail to reject H0 - no significant reduction in screen time.")
```
