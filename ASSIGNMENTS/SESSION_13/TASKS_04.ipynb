#### Task 4: Compare average ratings of Zomato vs Swiggy (15 users each) with Welch's two-sample t-test.

```python
import numpy as np
import pandas as pd
from scipy import stats
 
alpha = 0.05
zomato = np.array([4.2, 3.8, 4.5, 4.0, 3.9, 4.4, 4.1, 3.7, 4.3, 4.6, 3.8, 4.0, 4.2, 4.4, 3.9])
swiggy = np.array([4.0, 3.5, 4.1, 3.8, 3.6, 4.2, 3.9, 3.4, 4.0, 4.3, 3.7, 3.6, 4.1, 3.9, 3.5])
 
print("H0: mean rating of Zomato = mean rating of Swiggy")
print("H1: mean rating of Zomato != mean rating of Swiggy   (two-tailed)\n")
 
print(pd.DataFrame({"n": [15, 15], "mean": [zomato.mean(), swiggy.mean()],
                    "std": [zomato.std(ddof=1), swiggy.std(ddof=1)]},
                   index=["Zomato", "Swiggy"]).round(3).to_string(), "\n")
 
# Assumption checks
print(f"Shapiro-Wilk normality: Zomato p = {stats.shapiro(zomato).pvalue:.3f}, "
      f"Swiggy p = {stats.shapiro(swiggy).pvalue:.3f}  (p > 0.05 -> roughly normal)")
print(f"Levene equal-variance test: p = {stats.levene(zomato, swiggy).pvalue:.3f}\n")
 
# Independent samples, unknown population variances -> Welch's two-sample t-test
t_stat, p_value = stats.ttest_ind(zomato, swiggy, equal_var=False)
print(f"Welch's t-test: t = {t_stat:.3f}, p-value = {p_value:.4f}")
print(f"Difference in means = {zomato.mean() - swiggy.mean():.3f} rating points")
 
if p_value < alpha:
    print("Decision: reject H0 - the average ratings of the two apps are significantly different.")
else:
    print("Decision: fail to reject H0 - no significant difference between the average ratings.")
 
# Robustness check because ratings are ordinal and n is small
u_stat, u_p = stats.mannwhitneyu(zomato, swiggy, alternative="two-sided")
print(f"\nRobustness check (Mann-Whitney U): U = {u_stat:.1f}, p-value = {u_p:.4f}")
```
