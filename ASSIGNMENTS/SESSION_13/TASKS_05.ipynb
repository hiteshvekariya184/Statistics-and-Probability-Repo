#### Task 5: One-sample t-test on average daily UPI transactions per user (PhonePe-style data)
Code written with an AI assistant (Claude), then run on my own simulated dataset.
Prompt: "Write Python code that performs a one-sample t-test on a dataset such as the
average daily UPI transactions per user on PhonePe."

```python
import numpy as np
import pandas as pd
from scipy import stats
 
rng = np.random.default_rng(seed=5)
 
# Dataset: average daily UPI transactions for 40 users (simulated)
transactions = rng.normal(loc=4.6, scale=1.4, size=40).round(1).clip(min=0.5)
claimed_mean = 4.0            # company claims the average user makes 4 UPI payments a day
alpha = 0.05
 
print("H0: mean daily UPI transactions per user = 4.0")
print("H1: mean daily UPI transactions per user != 4.0   (two-tailed)\n")
print("Data:", transactions.tolist(), "\n")
 
s = pd.Series(transactions)
print(s.describe().round(2).to_string(), "\n")
 
t_stat, p_value = stats.ttest_1samp(transactions, popmean=claimed_mean)
se = s.std(ddof=1) / np.sqrt(len(s))
ci = stats.t.interval(0.95, len(s) - 1, loc=s.mean(), scale=se)
 
print(f"t-statistic = {t_stat:.3f}, degrees of freedom = {len(s) - 1}")
print(f"p-value     = {p_value:.4f}")
print(f"95% CI for the mean = ({ci[0]:.2f}, {ci[1]:.2f})")
 
if p_value < alpha:
    print("\nDecision: reject H0 - the average is significantly different from 4.0 transactions/day.")
else:
    print("\nDecision: fail to reject H0 - no significant difference from 4.0 transactions/day.")
```
