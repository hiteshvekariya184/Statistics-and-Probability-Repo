#### Task 2: Independent two-sample t-test - daily steps before vs after installing a fitness app.sks

```python
import numpy as np
import pandas as pd
from scipy import stats
 
rng = np.random.default_rng(seed=2)
alpha = 0.05
 
# Two DIFFERENT groups of users (simulated): 40 users without the app, 40 users with it
before = rng.normal(loc=6500, scale=1500, size=40).round()
after = rng.normal(loc=7400, scale=1600, size=40).round()
 
print("H0: mean steps (after app) <= mean steps (before app)")
print("H1: mean steps (after app) >  mean steps (before app)   (one-tailed)\n")
 
summary = pd.DataFrame({"n": [len(before), len(after)],
                        "mean": [before.mean(), after.mean()],
                        "std": [before.std(ddof=1), after.std(ddof=1)]},
                       index=["Before app", "After app"]).round(1)
print(summary.to_string(), "\n")
 
# Check equal variances (Levene) to decide between the pooled and Welch version
lev_stat, lev_p = stats.levene(before, after)
equal_var = lev_p > 0.05
print(f"Levene test for equal variances: p = {lev_p:.3f} -> "
      f"{'variances similar (pooled t-test)' if equal_var else 'variances differ (use Welch t-test)'}")
 
t_stat, p_value = stats.ttest_ind(after, before, equal_var=equal_var, alternative="greater")
print(f"\nttest_ind(after, before, alternative='greater'): t = {t_stat:.3f}, p-value = {p_value:.5f}")
print(f"Average increase = {after.mean() - before.mean():.0f} steps per day")
 
if p_value < alpha:
    print("Decision: reject H0 - average daily steps increased significantly after using the app.")
else:
    print("Decision: fail to reject H0 - no significant increase in daily steps.")
```
