#### Task 2: Z-score of a 9500-step day (mean 8000, SD 1200).

```python
import numpy as np
from scipy.stats import norm
 
mu, sigma = 8000, 1200
x = 9500
 
# Z = (X - mu) / sigma
z = (x - mu) / sigma
print(f"Z = (X - mu) / sigma = ({x} - {mu}) / {sigma} = {z:.2f}")
print(f"The 9500-step day is {z:.2f} standard deviations above the mean.")
print(f"Percentile: {norm.cdf(z):.1%} of days have fewer steps than this day "
      f"(only {norm.sf(z):.1%} of days are higher).\n")
 
# The same formula applied to a whole array of daily step counts (14 days)
steps = np.array([7200, 8100, 6500, 9500, 8800, 7600, 10200, 5900, 8400, 9100, 7000, 8000, 9900, 6800])
z_scores = (steps - mu) / sigma
print("Day  Steps   Z-score")
for day, (s, zz) in enumerate(zip(steps, z_scores), start=1):
    flag = "  <-- the 9500-step day" if s == x else ""
    print(f"{day:>3}  {s:>5}   {zz:>6.2f}{flag}")
 
print(f"\nCheck: the 9500-step row in the table has Z = {z_scores[steps == x][0]:.2f}, matching the single calculation above.")
print("Rule of thumb: |Z| > 2 is unusual, |Z| > 3 is very unusual. Z = 1.25 is a good but ordinary day.")
```
