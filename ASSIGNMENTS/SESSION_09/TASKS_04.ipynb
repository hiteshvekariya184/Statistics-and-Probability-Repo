#### Task 4: Probability that a Zomato delivery takes less than 25 minutes (mean 30, SD 5).

```python
from scipy.stats import norm
 
mu, sigma = 30, 5
 
# P(X < 25) with scipy.stats.norm.cdf
p = norm.cdf(25, loc=mu, scale=sigma)
z = (25 - mu) / sigma
print(f"Z for 25 minutes = (25 - {mu}) / {sigma} = {z:.2f}")
print(f"P(delivery time < 25 min) = norm.cdf(25, {mu}, {sigma}) = {p:.4f}  ({p:.2%})")
print(f"Check with the standard normal: norm.cdf({z:.0f}) = {norm.cdf(z):.4f}\n")
 
print("A few related probabilities:")
print(f"  P(time > 35 min)        = {norm.sf(35, mu, sigma):.4f}")
print(f"  P(25 < time < 35 min)   = {norm.cdf(35, mu, sigma) - norm.cdf(25, mu, sigma):.4f}")
print(f"  Time that 90% of deliveries beat = {norm.ppf(0.90, mu, sigma):.1f} minutes")
```
