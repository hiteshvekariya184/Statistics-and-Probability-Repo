#### Task 5: Supporting demo for 'why the normal distribution matters' in app analytics.

```python
import numpy as np
from scipy import stats
 
rng = np.random.default_rng(seed=5)
 
# 1) The 68-95-99.7 rule on delivery times (normal, mean 30, SD 5)
times = rng.normal(30, 5, 100_000)
print("1) Empirical rule on simulated delivery times (mean 30, SD 5):")
for k, theory in [(1, 68.27), (2, 95.45), (3, 99.73)]:
    share = np.mean(np.abs(times - 30) <= k * 5) * 100
    print(f"   within {k} SD ({30 - 5 * k}-{30 + 5 * k} min): {share:.2f}%  (theory {theory}%)")
 
# 2) Z-scores flag unusual orders
sample = np.array([28, 31, 29, 33, 27, 30, 52, 32, 29, 31])
z = (sample - 30) / 5
print("\n2) Z-scores flag unusual delivery times (|Z| > 3):")
for t, zz in zip(sample, z):
    print(f"   {t:>3} min -> Z = {zz:>5.1f}{'  <-- unusual, investigate' if abs(zz) > 3 else ''}")
 
# 3) Averages become normal even when the raw data is skewed (Central Limit Theorem)
orders = rng.lognormal(mean=6, sigma=0.6, size=200_000)           # skewed order values
means = rng.choice(orders, size=(5000, 40)).mean(axis=1)           # 5000 samples of 40 orders
print("\n3) Skewed order values vs the averages of 40 orders:")
print(f"   skewness of raw order values = {stats.skew(orders):.2f}")
print(f"   skewness of sample averages  = {stats.skew(means):.2f}  (much closer to 0, the normal shape)")
print("   This is why confidence intervals and hypothesis tests on averages work for sales data.")
```
