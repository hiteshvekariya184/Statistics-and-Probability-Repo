#### Task 3: Standard normal curve with the area between Z = -1 and Z = 1 shaded.

```python
import os
import numpy as np
import matplotlib
if "__file__" in globals():
    matplotlib.use("Agg")
import matplotlib.pyplot as plt
from scipy.stats import norm
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
 
z = np.linspace(-4, 4, 500)
pdf = norm.pdf(z, loc=0, scale=1)
 
area = norm.cdf(1) - norm.cdf(-1)              # P(-1 < Z < 1)
print(f"P(-1 < Z < 1) = cdf(1) - cdf(-1) = {norm.cdf(1):.4f} - {norm.cdf(-1):.4f} = {area:.4f} ({area:.2%})")
 
plt.figure(figsize=(8, 4.5))
plt.plot(z, pdf, color="black", lw=2)
mask = (z >= -1) & (z <= 1)
plt.fill_between(z[mask], pdf[mask], color="#2a6f97", alpha=0.6, label=f"P(-1 < Z < 1) = {area:.2%}")
plt.axvline(-1, color="red", ls="--")
plt.axvline(1, color="red", ls="--")
plt.text(0, 0.17, f"{area:.2%}", ha="center", fontsize=14, color="white", fontweight="bold")
plt.xlabel("Z (standard deviations from the mean)")
plt.ylabel("Probability density")
plt.title("Standard normal curve: area within one standard deviation")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(HERE, "standard_normal.png"), dpi=150)
if "__file__" not in globals():
    plt.show()
print("Plot saved to standard_normal.png")
```
