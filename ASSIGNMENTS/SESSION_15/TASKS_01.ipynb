#### Task 1: Create purchase_data.csv (15 customers, 5 per age group).

```python
import os
import pandas as pd
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
 
# Sample data (illustrative): purchase_value in rupees
data = {
    "customer_id": [f"C{i:03d}" for i in range(1, 16)],
    "age_group": ["18-25"] * 5 + ["26-35"] * 5 + ["36-50"] * 5,
    "purchase_value": [450, 520, 480, 610, 540,      # 18-25
                       700, 650, 780, 720, 690,      # 26-35
                       640, 720, 590, 680, 660],     # 36-50
}
df = pd.DataFrame(data)
path = os.path.join(HERE, "purchase_data.csv")
df.to_csv(path, index=False)
 
print(f"Saved {len(df)} records to {path}\n")
print(df.to_string(index=False))
print("\nRecords per age group:")
print(df["age_group"].value_counts().sort_index().to_string())
```
