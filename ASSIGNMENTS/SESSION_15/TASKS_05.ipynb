#### Task 5: One-way ANOVA on Zomato monthly order amounts for three customer segments.
Code generated with an AI assistant (Claude), then run on dummy data.
Prompt: "Write Python code that performs a one-way ANOVA on the average monthly order amounts
of three Zomato customer segments: Students, Working Professionals and Retirees.

```python
import numpy as np
import pandas as pd
from scipy import stats
 
rng = np.random.default_rng(seed=5)
 
# Dummy data: average monthly order amount (Rs) for 20 customers in each segment
students = rng.normal(loc=900, scale=250, size=20).round()
professionals = rng.normal(loc=2200, scale=600, size=20).round()
retirees = rng.normal(loc=1300, scale=400, size=20).round()
 
orders = pd.DataFrame({
    "segment": ["Students"] * 20 + ["Working Professionals"] * 20 + ["Retirees"] * 20,
    "monthly_order_amount": np.concatenate([students, professionals, retirees]),
})
 
print("H0: average monthly order amount is the same for all three segments")
print("H1: at least one segment has a different average\n")
print(orders.groupby("segment")["monthly_order_amount"].agg(["count", "mean", "std"]).round(1).to_string(), "\n")
 
f_stat, p_value = stats.f_oneway(students, professionals, retirees)
print(f"F-statistic = {f_stat:.3f}")
print(f"p-value     = {p_value:.3g}\n")
 
alpha = 0.05
if p_value < alpha:
    print("Decision: reject H0 - average monthly order amounts differ significantly between segments.")
    res = stats.tukey_hsd(students, professionals, retirees)
    names = ["Students", "Working Professionals", "Retirees"]
    means = [students.mean(), professionals.mean(), retirees.mean()]
    print("\nTukey HSD post-hoc comparisons:")
    for i in range(3):
        for j in range(i + 1, 3):
            print(f"  {names[i]} vs {names[j]}: diff = {means[i] - means[j]:8.1f}, p = {res.pvalue[i][j]:.4f}")
else:
    print("Decision: fail to reject H0 - no significant difference between segments.")
```
