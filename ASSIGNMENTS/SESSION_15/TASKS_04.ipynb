#### Task 4: Between-group and within-group variance for purchase_value (the pieces of the ANOVA F-ratio).
Run task1_create_purchase_data.py first so purchase_data.csv exists.

```python
import os
import numpy as np
import pandas as pd
from scipy import stats
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
df = pd.read_csv(os.path.join(HERE, "purchase_data.csv"))
 
values = df["purchase_value"]
grand_mean = values.mean()
N = len(df)
k = df["age_group"].nunique()
g = df.groupby("age_group")["purchase_value"]
n_i, mean_i = g.count(), g.mean()
 
# Sum of squares
ss_between = (n_i * (mean_i - grand_mean) ** 2).sum()                       # group means vs grand mean
ss_within = ((df["purchase_value"] - g.transform("mean")) ** 2).sum()       # values vs own group mean
ss_total = ((values - grand_mean) ** 2).sum()
 
# Variances (mean squares) = sum of squares / degrees of freedom
df_between, df_within = k - 1, N - k
between_variance = ss_between / df_between      # MS between
within_variance = ss_within / df_within         # MS within
 
print(f"Grand mean = {grand_mean:.2f}")
print("Group means:", mean_i.round(2).to_dict(), "\n")
print(f"Sum of squares between (SSB) = {ss_between:.2f}   (df = {df_between})")
print(f"Sum of squares within  (SSW) = {ss_within:.2f}   (df = {df_within})")
print(f"Sum of squares total   (SST) = {ss_total:.2f}   (SSB + SSW = {ss_between + ss_within:.2f})\n")
 
print(f"Between-group variance (MSB = SSB / {df_between}) = {between_variance:.2f}")
print(f"Within-group variance  (MSW = SSW / {df_within}) = {within_variance:.2f}")
f_manual = between_variance / within_variance
print(f"F = MSB / MSW = {f_manual:.4f}")
 
f_scipy, p = stats.f_oneway(*[x.values for _, x in g])
print(f"scipy f_oneway check: F = {f_scipy:.4f}, p = {p:.6f}")
```
