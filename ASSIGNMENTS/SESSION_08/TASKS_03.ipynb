#### Task 3: Poisson probability of exactly 6 Spotify playlist creations in a minute (average rate = 4).

```python
from math import exp, factorial
from scipy.stats import poisson
 
lam, k = 4, 6
prob = poisson.pmf(k, lam)
print(f"P(exactly {k} playlists in a minute | lambda = {lam}) = {prob:.4f}  ({prob:.2%})")
 
# Manual check with the formula e^(-lambda) * lambda^k / k!
manual = exp(-lam) * lam ** k / factorial(k)
print(f"Manual check: e^-{lam} x {lam}^{k} / {k}! = {exp(-lam):.5f} x {lam ** k} / {factorial(k)} = {manual:.4f}")
 
print("\nProbabilities for 0 to 10 playlists per minute:")
for x in range(11):
    print(f"  P(X = {x:>2}) = {poisson.pmf(x, lam):.4f}")
print(f"\nP(X >= 7) = {poisson.sf(6, lam):.4f}   (more than 6 in a minute)")
```
