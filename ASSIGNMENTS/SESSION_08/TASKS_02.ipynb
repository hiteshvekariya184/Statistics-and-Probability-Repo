#### Task 2: Binomial probability of exactly 3 successful Paytm UPI payments in 5 attempts (p = 0.9).

```python
from math import comb
from scipy.stats import binom
 
 
def binomial_probability(k, n, p):
    """P(X = k) for X ~ Binomial(n, p), using scipy.stats.binom.pmf."""
    return binom.pmf(k, n, p)
 
 
n, k, p = 5, 3, 0.9
prob = binomial_probability(k, n, p)
print(f"P(exactly {k} successes out of {n} attempts, p = {p}) = {prob:.4f}  ({prob:.2%})")
 
# Manual check with the formula C(n, k) * p^k * (1 - p)^(n - k)
manual = comb(n, k) * p ** k * (1 - p) ** (n - k)
print(f"Manual check: C({n},{k}) x {p}^{k} x {1 - p:.1f}^{n - k} = {comb(n, k)} x {p ** k:.3f} x {(1 - p) ** (n - k):.2f} = {manual:.4f}")
 
print("\nFull distribution of successful payments out of 5:")
for x in range(n + 1):
    print(f"  P(X = {x}) = {binomial_probability(x, n, p):.4f}")
print(f"\nMean = n x p = {n * p}, variance = n x p x (1 - p) = {n * p * (1 - p):.2f}")
```
