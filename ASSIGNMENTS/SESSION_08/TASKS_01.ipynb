#### Task 1: Simulate Bernoulli trials - does a user log in to Zomato successfully? (p = 0.85)

```python
import numpy as np
 
p = 0.85                                    # probability of a successful login
n_trials = 20
rng = np.random.default_rng(seed=1)
 
# Each trial is a Bernoulli(p): 1 = success, 0 = failure
results = (rng.random(n_trials) < p).astype(int)
 
print(f"Bernoulli simulation: p = {p}, {n_trials} login attempts")
print("Results:", results.tolist())
for i, r in enumerate(results, start=1):
    print(f"  Attempt {i:>2}: {'success (1)' if r == 1 else 'failure (0)'}")
 
print(f"\nSuccesses = {results.sum()}, failures = {n_trials - results.sum()}")
print(f"Observed success rate = {results.mean():.2f} (theoretical p = {p})")
 
# Optional: the same experiment with scipy's Bernoulli distribution
from scipy import stats
results2 = stats.bernoulli.rvs(p, size=n_trials, random_state=1)
print("\nscipy.stats.bernoulli.rvs gives:", results2.tolist())
```
