#### Task 5: Poisson use-case - Swiggy orders arriving at a restaurant, used for kitchen staffing.
Use-case text written with an AI assistant (Claude); this code turns it into numbers.

```python
import numpy as np
from scipy.stats import poisson
 
lam = 6                      # average orders arriving at the restaurant per 10-minute window
print(f"Poisson use-case: Swiggy orders per 10-minute window, lambda = {lam}\n")
 
# Question 1: how likely is a rush of more than 10 orders in a 10-minute window?
p_rush = poisson.sf(10, lam)          # P(X > 10)
print(f"P(more than 10 orders in 10 minutes) = {p_rush:.4f} ({p_rush:.1%})")
 
# Question 2: how many orders should the kitchen be ready for to cover 95% of windows?
capacity = int(poisson.ppf(0.95, lam))
print(f"Prepare for up to {capacity} orders per window to cover 95% of windows "
      f"(P(X <= {capacity}) = {poisson.cdf(capacity, lam):.3f})")
 
# Question 3: how many quiet windows (2 orders or fewer)?
p_quiet = poisson.cdf(2, lam)
print(f"P(2 or fewer orders) = {p_quiet:.4f} ({p_quiet:.1%})")
 
# Simulate one day of the dinner shift: 36 windows of 10 minutes (6 hours)
rng = np.random.default_rng(seed=5)
day = rng.poisson(lam, size=36)
print(f"\nSimulated 36 windows: mean = {day.mean():.2f} orders, busiest window = {day.max()}, "
      f"windows above capacity {capacity}: {(day > capacity).sum()}")
```
