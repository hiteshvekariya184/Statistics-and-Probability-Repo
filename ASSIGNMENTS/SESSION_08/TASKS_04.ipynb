#### Task 4: Simulate and plot Flipkart review counts in 12 random hours (Poisson, average 10 per hour)

```python
import os
import numpy as np
import matplotlib
if "__file__" in globals():            # plain script: save to file; notebook: show inline
    matplotlib.use("Agg")
import matplotlib.pyplot as plt
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
 
lam, n_hours = 10, 12
np.random.seed(4)
reviews = np.random.poisson(lam=lam, size=n_hours)      # reviews received in each of 12 hours
 
print(f"Poisson simulation: average rate = {lam} reviews/hour, {n_hours} randomly selected hours")
for h, r in enumerate(reviews, start=1):
    print(f"  Hour {h:>2}: {r} reviews")
print(f"\nSample mean = {reviews.mean():.2f} (theoretical {lam}), sample variance = {reviews.var(ddof=1):.2f} (theoretical {lam})")
print(f"Minimum = {reviews.min()}, maximum = {reviews.max()}")
 
plt.figure(figsize=(8, 4.5))
plt.bar(range(1, n_hours + 1), reviews, color="#2a6f97", zorder=3, label="Reviews in that hour")
plt.axhline(lam, color="red", ls="--", lw=2, label=f"Average rate = {lam}")
for h, r in enumerate(reviews, start=1):
    plt.text(h, r + 0.2, str(r), ha="center")
plt.xticks(range(1, n_hours + 1))
plt.xlabel("Randomly selected hour")
plt.ylabel("Number of reviews")
plt.title("Flipkart product reviews per hour (Poisson simulation)")
plt.legend()
plt.grid(axis="y", alpha=0.3, zorder=0)
plt.tight_layout()
plt.savefig(os.path.join(HERE, "reviews_poisson.png"), dpi=150)
if "__file__" not in globals():
    plt.show()
print("Plot saved to reviews_poisson.png")
```
