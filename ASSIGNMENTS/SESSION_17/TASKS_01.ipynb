#### Task 1: Identify the independent and dependent variable (YouTube influencers).

```python
import os
import pandas as pd
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
 
# Illustrative dataset of 10 anonymised YouTube influencers
influencers = pd.DataFrame({
    "influencer": [f"Creator {i:02d}" for i in range(1, 11)],
    "followers_thousands": [50, 120, 200, 350, 500, 800, 1200, 1800, 2500, 4000],
    "avg_monthly_brand_deals": [1, 1, 2, 2, 3, 4, 5, 6, 8, 10],
})
influencers.to_csv(os.path.join(HERE, "influencers.csv"), index=False)
print(influencers.to_string(index=False), "\n")
 
independent = "followers_thousands"
dependent = "avg_monthly_brand_deals"
print(f"Independent variable (X, predictor): {independent}")
print(f"Dependent variable   (Y, outcome)  : {dependent}\n")
 
# How does Y change as X changes? (quick check of the direction of the relationship)
r = influencers[independent].corr(influencers[dependent])
print(f"Correlation between X and Y = {r:.3f}  -> more followers go with more brand deals")
print("Reasoning: brands choose creators for their reach, so the number of brand deals DEPENDS on the "
      "number of followers, not the other way round.")
```
