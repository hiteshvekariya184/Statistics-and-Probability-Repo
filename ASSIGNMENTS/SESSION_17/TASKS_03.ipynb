#### Task 3: Simple linear regression (scikit-learn): Swiggy daily orders vs daily advertising spend.

```python
import os
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
rng = np.random.default_rng(seed=3)
 
# Simulated data for 15 days: ad spend in lakhs, orders per day
spend = rng.integers(2, 17, size=15)
orders = (12000 + 1800 * spend + rng.normal(0, 2500, size=15)).round().astype(int)
swiggy = pd.DataFrame({"ad_spend_lakhs": spend, "daily_orders": orders}).sort_values("ad_spend_lakhs")
swiggy.to_csv(os.path.join(HERE, "swiggy_ads.csv"), index=False)
print(swiggy.to_string(index=False), "\n")
 
X = swiggy[["ad_spend_lakhs"]]          # scikit-learn needs a 2-D feature table
y = swiggy["daily_orders"]
 
model = LinearRegression()
model.fit(X, y)
 
print(f"Slope (coefficient) = {model.coef_[0]:.2f} orders per extra lakh of ad spend")
print(f"Intercept           = {model.intercept_:.2f} orders per day when ad spend is 0")
print(f"\nModel: daily_orders = {model.intercept_:.0f} + {model.coef_[0]:.0f} x ad_spend_lakhs")
 
print("\nExample predictions:")
for s in (5, 10, 20):
    print(f"  Ad spend {s:>2} lakhs -> about {model.predict(pd.DataFrame({'ad_spend_lakhs': [s]}))[0]:,.0f} orders/day")
```
