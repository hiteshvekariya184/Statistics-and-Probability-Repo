#### Task 4: R-squared of the Swiggy regression model.
Run task3_swiggy_linear_regression.py first so swiggy_ads.csv exists.

```python
import os
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
 
HERE = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
swiggy = pd.read_csv(os.path.join(HERE, "swiggy_ads.csv"))
X, y = swiggy[["ad_spend_lakhs"]], swiggy["daily_orders"]
 
model = LinearRegression().fit(X, y)
pred = model.predict(X)
 
r2 = r2_score(y, pred)                                  # same as model.score(X, y)
print(f"R-squared (r2_score)   = {r2:.4f}")
print(f"R-squared (model.score) = {model.score(X, y):.4f}")
 
# Manual calculation: R^2 = 1 - SSE / SST
sse = ((y - pred) ** 2).sum()              # unexplained variation (residuals)
sst = ((y - y.mean()) ** 2).sum()          # total variation around the mean
print(f"\nSSE (unexplained) = {sse:,.0f}")
print(f"SST (total)       = {sst:,.0f}")
print(f"R-squared = 1 - SSE/SST = 1 - {sse / sst:.4f} = {1 - sse / sst:.4f}")
print(f"Correlation r = {np.sqrt(r2):.4f}  (R-squared is r squared for simple regression)")
print(f"Average prediction error (RMSE) = {np.sqrt(sse / len(y)):,.0f} orders/day")
 
print(f"\nInterpretation: R-squared = {r2:.2f}, which is close to 1. About {r2:.0%} of the variation in daily "
      f"orders is explained by ad spend; the remaining {1 - r2:.0%} is due to other factors and random noise.")
```
