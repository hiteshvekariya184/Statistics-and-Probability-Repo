#### Task 3: Interquartile range (IQR) of a month of daily step counts and what it says about consistency.

```python
import numpy as np
 
rng = np.random.default_rng(seed=3)
steps = rng.normal(loc=8000, scale=1500, size=30).round().astype(int)      # 30 days of step counts (simulated)
print("Daily steps (30 days):", steps.tolist(), "\n")
 
q1, q3 = np.percentile(steps, [25, 75])
iqr = q3 - q1                                    # IQR = Q3 - Q1
median = np.median(steps)
print(f"Q1 = {q1:,.1f}")
print(f"Q3 = {q3:,.1f}")
print(f"IQR = Q3 - Q1 = {q3:,.1f} - {q1:,.1f} = {iqr:,.1f} steps")
print(f"Median = {median:,.1f}; IQR as a share of the median = {iqr / median:.1%}")
print(f"For comparison: range = {steps.max() - steps.min():,} steps, standard deviation = {steps.std(ddof=1):,.0f}\n")
 
print("Interpretation:")
print(f"  The middle 50% of days fall within a band of {iqr:,.0f} steps (between {q1:,.0f} and {q3:,.0f}).")
if iqr / median < 0.15:
    print("  The band is narrow compared with the median, so daily activity is quite consistent.")
elif iqr / median < 0.30:
    print("  The band is moderate compared with the median, so activity is fairly consistent with some ups and downs.")
else:
    print("  The band is wide compared with the median, so activity is inconsistent from day to day.")
print("  A smaller IQR means steadier routine; a larger IQR means more variable days (for example active weekends and lazy weekdays).")
```
