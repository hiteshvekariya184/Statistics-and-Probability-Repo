#### Task 2: Divide IPL player scores into quartiles with numpy.quantile and print Q1, Q2 (median) and Q3.

```python
import numpy as np
 
scores = np.array([12, 45, 78, 33, 5, 61, 90, 24, 52, 17, 68, 39, 101, 8, 47])   # runs scored by 15 players
print("IPL player scores:", scores.tolist())
print("Sorted           :", np.sort(scores).tolist(), "\n")
 
q1, q2, q3 = np.quantile(scores, [0.25, 0.50, 0.75])
print(f"Q1 (25th percentile) = {q1:.2f}")
print(f"Q2 (median, 50th)    = {q2:.2f}")
print(f"Q3 (75th percentile) = {q3:.2f}")
print(f"Minimum = {scores.min()}, Maximum = {scores.max()}\n")
 
# Divide the players into the four quartile groups
group = np.digitize(scores, [q1, q2, q3], right=True) + 1       # 1 = lowest quarter ... 4 = highest quarter
names = {1: "Q1 group (lowest 25%)", 2: "Q2 group", 3: "Q3 group", 4: "Q4 group (highest 25%)"}
for g in (1, 2, 3, 4):
    members = np.sort(scores[group == g]).tolist()
    print(f"{names[g]:<24}: {members}")
```
