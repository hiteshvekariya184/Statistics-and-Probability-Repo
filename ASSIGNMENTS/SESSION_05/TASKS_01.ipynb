#### Task 1: 25th, 50th and 75th percentiles of 20 Zomato delivery times using numpy.percentile.

```python
import numpy as np
 
delivery_times = [25, 30, 28, 35, 40, 22, 45, 32, 29, 38, 27, 33, 50, 31, 26, 36, 34, 29, 41, 30]   # minutes
print(f"Delivery times ({len(delivery_times)} orders):", delivery_times)
print("Sorted:", sorted(delivery_times), "\n")
 
p25, p50, p75 = np.percentile(delivery_times, [25, 50, 75])
print(f"25th percentile = {p25:.2f} minutes")
print(f"50th percentile = {p50:.2f} minutes  (the median)")
print(f"75th percentile = {p75:.2f} minutes\n")
 
print("Meaning:")
print(f"  25% of deliveries took {p25:.2f} minutes or less")
print(f"  50% of deliveries took {p50:.2f} minutes or less")
print(f"  75% of deliveries took {p75:.2f} minutes or less (only 25% took longer)")
print(f"\nCheck with the mean: {np.mean(delivery_times):.2f} minutes; numpy.median gives {np.median(delivery_times):.2f}")
```
