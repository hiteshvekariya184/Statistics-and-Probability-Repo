#### Task 4: Detect outliers in Swiggy order amounts with the IQR method.

```python
import numpy as np
 
def detect_outliers_iqr(values, k=1.5):
    """Return the outliers, using the rule: value < Q1 - k*IQR or value > Q3 + k*IQR."""
    values = np.asarray(values)
    q1, q3 = np.percentile(values, [25, 75])
    iqr = q3 - q1
    lower, upper = q1 - k * iqr, q3 + k * iqr
    outliers = values[(values < lower) | (values > upper)]
    return outliers, {"Q1": q1, "Q3": q3, "IQR": iqr, "lower_fence": lower, "upper_fence": upper}
 
 
order_amounts = [250, 320, 280, 450, 300, 380, 260, 310, 290, 2200, 330, 270, 350, 310, 295,
                 3100, 340, 305, 285, 360, 275, 320, 40, 315, 300]          # rupees
print("Swiggy order amounts:", order_amounts, "\n")
 
outliers, stats_ = detect_outliers_iqr(order_amounts)
print(f"Q1 = {stats_['Q1']:.1f}, Q3 = {stats_['Q3']:.1f}, IQR = {stats_['IQR']:.1f}")
print(f"Lower fence = Q1 - 1.5 x IQR = {stats_['lower_fence']:.1f}")
print(f"Upper fence = Q3 + 1.5 x IQR = {stats_['upper_fence']:.1f}\n")
print(f"Outlier values: {sorted(outliers.tolist())}")
print(f"Number of outliers: {len(outliers)} of {len(order_amounts)} orders")
 
clean = [x for x in order_amounts if x not in outliers]
print(f"\nMean with outliers = {np.mean(order_amounts):.1f}; mean without outliers = {np.mean(clean):.1f}")
print("Outliers distort the mean, which is why the median and IQR are preferred for skewed data.")
```
