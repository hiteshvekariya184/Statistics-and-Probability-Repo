## Task 4: Mean vs Median of UPI Transactions

```python
import numpy as np
import pandas as pd
 
upi_amounts = pd.Series([100, 120, 105, 110, 5000, 115, 108], name="upi_amount")
 
mean_amt = upi_amounts.mean()
median_amt = upi_amounts.median()
 
print("Sorted amounts:", upi_amounts.sort_values().tolist())
print(f"Mean   (pandas) : Rs.{mean_amt:.2f}")
print(f"Median (pandas) : Rs.{median_amt:.0f}")
print(f"Mean   (numpy)  : Rs.{np.mean(upi_amounts):.2f}")
print(f"Median (numpy)  : Rs.{np.median(upi_amounts):.0f}")
 
print()
print("Explanation:")
print("The median (Rs.110) is the better measure of a 'typical' transaction.")
print("The single Rs.5000 outlier pulls the mean up to Rs.808.29, even though")
print("six of the seven payments are between Rs.100 and Rs.120. The median")
print("depends only on the middle position, so extreme values do not affect it.")
```
