## Task 2: Median Delivery Time

```python
import numpy as np
import pandas as pd
 
 
def find_median(data):
    """Return the median of a list using NumPy."""
    return np.median(data)
 
 
delivery_times = [32, 28, 29, 45, 30, 31, 60, 30, 29]
 
s = pd.Series(delivery_times, name="delivery_minutes")
 
print("Sorted times            :", sorted(delivery_times))
print("Median (function/numpy) :", find_median(delivery_times), "minutes")
print("Median (pandas)         :", s.median(), "minutes")
```
