## Task 1: Mean of 7 Daily Step Counts

```python
import numpy as np
import pandas as pd
 
# Daily step counts for 7 days (Mon-Sun). Replace with your own data.
steps = pd.Series(
    [8234, 10567, 6789, 12045, 9321, 7456, 11230],
    index=["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    name="steps",
)
 
print(steps)
print()
print(f"Total steps (pandas) : {steps.sum()}")
print(f"Mean steps (pandas)  : {steps.mean():.2f}")
print(f"Mean steps (numpy)   : {np.mean(steps.to_numpy()):.2f}")
```
