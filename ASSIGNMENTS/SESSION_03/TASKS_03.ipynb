## Task 3: Mode of YouTube Video Genres

```python
import pandas as pd
 
genres = ['Music', 'Vlog', 'Music', 'Tech', 'Music', 'Vlog',
          'Tech', 'Music', 'Comedy', 'Vlog', 'Music', 'Comedy']
 
s = pd.Series(genres, name="genre")
 
# Count of each genre
counts = s.value_counts()
print("Genre counts:")
print(counts)
print()
 
# Mode = most frequent value
mode_genre = s.mode()[0]
print(f"Mode (most watched genre): {mode_genre} ({counts[mode_genre]} videos)")
```
